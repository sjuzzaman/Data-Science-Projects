# Virginia Carbon Monoxide Analysis: Ground Monitoring, Weather, and Satellite Observations

Analyze carbon monoxide levels across Virginia from 2020–2026 using EPA ground-monitoring data, NOAA weather observations, and NASA satellite measurements. The project examines spatial and seasonal pollution patterns, evaluates relationships between meteorological conditions and CO concentrations, and compares satellite observations with ground-level measurements. The analysis includes data cleaning, quality control, geospatial matching, exploratory data analysis, visualization, statistical modeling, and time-aware model evaluation.

## Dataset Column Descriptions

The seven annual CO datasets (`CO Virginia 2020.csv` through `CO Virginia 2026.csv`) use the same EPA/AQS column structure. Each row represents one daily carbon monoxide summary for one monitoring site.

| Column | Meaning |
|---|---|
| `Date` | Observation date for the daily summary. |
| `Source` | Data source; `AQS` refers to EPA's Air Quality System. |
| `Site ID` | Unique EPA monitoring site identifier. |
| `POC` | Parameter Occurrence Code, used to distinguish multiple instruments measuring the same pollutant at the same site. |
| `Daily Max 8-hour CO Concentration` | Highest rolling 8-hour average carbon monoxide concentration measured that day. This is the main CO pollution measurement. |
| `Units` | Measurement unit for CO concentration, usually `ppm`. |
| `Daily AQI Value` | Air Quality Index value calculated from the daily CO concentration. |
| `Local Site Name` | Human-readable name of the monitoring site. |
| `Daily Obs Count` | Number of hourly observations available for that day. |
| `Percent Complete` | Percent of expected observations available for the day; useful for data quality checks. |
| `AQS Parameter Code` | EPA pollutant code; `42101` represents carbon monoxide. |
| `AQS Parameter Description` | Pollutant name or description, here `Carbon monoxide`. |
| `Method Code` | Code for the measurement method or instrument used. |
| `CBSA Code` | Core-Based Statistical Area code for the metro or micropolitan area. |
| `CBSA Name` | Name of the metro or micropolitan statistical area. |
| `State FIPS Code` | Federal numeric state code; Virginia is `51`. |
| `State` | State name. |
| `County FIPS Code` | Federal numeric county code. |
| `County` | County name. |
| `Site Latitude` | Latitude coordinate of the monitoring site. |
| `Site Longitude` | Longitude coordinate of the monitoring site. |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
CO_Virginia_2020 = pd.read_csv('CO Virginia 2020.csv')
CO_Virginia_2021 = pd.read_csv('CO Virginia 2021.csv')
CO_Virginia_2022 = pd.read_csv('CO Virginia 2022.csv')
CO_Virginia_2023 = pd.read_csv('CO Virginia 2023.csv')
CO_Virginia_2024 = pd.read_csv('CO Virginia 2024.csv')
CO_Virginia_2025 = pd.read_csv('CO Virginia 2025.csv')
CO_Virginia_2026 = pd.read_csv('CO Virginia 2026.csv')
```

```python
files = {
    2020: "CO Virginia 2020.csv",
    2021: "CO Virginia 2021.csv",
    2022: "CO Virginia 2022.csv",
    2023: "CO Virginia 2023.csv",
    2024: "CO Virginia 2024.csv",
    2025: "CO Virginia 2025.csv",
    2026: "CO Virginia 2026.csv",
}

dfs = []

for year, file in files.items():
    temp = pd.read_csv(file)
    temp["Year"] = year
    dfs.append(temp)

epa_df = pd.concat(dfs, ignore_index=True)
```

```python
epa_df["Date"] = pd.to_datetime(epa_df["Date"])
epa_df["Month"] = epa_df["Date"].dt.month
epa_df["Month Name"] = epa_df["Date"].dt.month_name()
```

```python
county_year_co = (epa_df.groupby(["County", "Year"])['Daily Max 8-hour CO Concentration'].mean().reset_index())

plt.figure(figsize=(14, 8))

for county in county_year_co["County"].unique():
    county_data = county_year_co[county_year_co['County'] == county]
    plt.plot(county_data["Year"], county_data["Daily Max 8-hour CO Concentration"], marker="o", label=f"CO in {county}")


plt.title("Average Carbon Monoxide Levels by County, 2020-2026")
plt.xlabel("Year")
plt.ylabel("Average CO Concentration (ppm)")
plt.legend()
plt.grid(True)
plt.show()
```
